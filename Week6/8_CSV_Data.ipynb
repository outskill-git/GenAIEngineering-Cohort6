# Notebook 8 — CSV Data Retrieval with Pandas and LLM Integration

## What You Will Learn

Not all data lives in a database. This notebook shows you three progressively powerful ways to query CSV files and use the results as LLM context — from manual pandas filtering, through LLM-generated pandas code, to running SQL directly on a DataFrame.

### Three Retrieval Approaches

| Approach | Tool | Best For |
|---|---|---|
| **Manual Pandas Filtering** | `DataFrame` boolean conditions | Known, fixed queries |
| **LLM-Generated Pandas Code** | Groq → `exec()` | Dynamic, natural-language queries |
| **PandasSQL** | `pandasql.sqldf()` | SQL-familiar users, reuse of SQL patterns |

### Topics Covered

- Loading CSV files and inspecting data types, min/max/mean, and unique values with `pandas`
- Writing boolean filter expressions and column slicing
- Prompting an LLM to generate pandas filter code from a question, then executing it safely
- Converting filtered DataFrames to CSV strings for compact LLM context
- Running SQL queries on DataFrames using `pandasql`

### Skills You Will Build

- Profile a dataset with `dtypes`, `min()`, `max()`, `mean()`, `unique()`
- Build multi-condition pandas filters and select specific columns
- Use an LLM as a code generator for data retrieval (prompt → code → exec → context)
- Pass filtered DataFrame content as CSV to an LLM for question answering
- Query DataFrames with SQL syntax using `pandasql`

> **Why it matters:** CSV files are the most common data format outside of databases. Mastering pandas-based retrieval gives you a fast, dependency-light RAG layer for any tabular data source.

**Use as Context**  
The data filtered using pandas used as context to provide LLM the data  
This data retrieval can be quick and effective  


```python
load_dotenv()
HF_Model = "google/flan-t5-base"
```

>Specific instructions in terms of Response

```python
R_Instr = "Using the context given, provide response to the user question or statement.\
            Context is provided as CSV formatted string.\
            Answer to the question with details"
```

```python
# User prompt
Prompt = "Why do you think students score low marks despite attending class and studying well?"

Filtered = Data[(Data['study_hours_per_day'] >= Data['study_hours_per_day'].mean()) & 
                (Data['attendance_percentage'] >= Data['attendance_percentage'].mean()) & 
                (Data['exam_score'] < 60)]
Context = Filtered.to_csv (index=False, float_format='%.1f')

# Invoke LLM with prompt and context
messages=[
    {
        "role": "system",
        "content": R_Instr
    },

    {
        "role": "user",
        "content":"Context : \n"+ Context + "Query : \n" + Prompt
    }
]
completion = hf_chat_completion(
    messages=messages,
    model="openai/gpt-oss-120b",
)

print (completion.choices[0].message.content)
```
